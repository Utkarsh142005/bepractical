```python
fruits = ["apple","banana","mango","orange"]
print(fruits)
print(len(fruits))
fruits.append("grapes")
more_fruits=["pineapple","cherry"]
fruits.extend(more_fruits)
print(fruits)
print("fruit 1 ", fruits[0])
print("last fruit", fruits[-1])
print(fruits[2:5])

fruits[1]="strawberry"
print("after updating",fruits)
fruits.remove("mango")
print("after removing mango",fruits)
removed__fruit = fruits.pop(2)
print(removed__fruit)

print("after pop",fruits)
fruits.sort()
print("after sort", fruits)
fruits.reverse()
print("after reversing",fruits)
fruits.clear()
print(fruits)
```
